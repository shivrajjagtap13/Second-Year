```python
import mysql.connector

# Database connection
mydb = mysql.connector.connect(
    host="localhost",
    user="root",
    password="shiv_n0t_f0und",
    database="studentdb"
)

print("Database connected successfully")

mycursor = mydb.cursor()

while True:
    print("\n----- MENU -----")
    print("1. Insert Record")
    print("2. Update Record")
    print("3. Delete Record")
    print("4. Display Records")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    # -------------------------------
    # INSERT
    # -------------------------------
    if choice == 1:
        id = int(input("Enter Student ID: "))
        name = input("Enter Student Name: ")
        age = int(input("Enter Age: "))
        course = input("Enter Department: ")

        sql = "INSERT INTO student1 (id, student_name, age, department) VALUES (%s, %s, %s, %s)"
        values = (id, name, age, course)

        mycursor.execute(sql, values)
        mydb.commit()

        print("Record inserted successfully")

    # -------------------------------
    # UPDATE
    # -------------------------------
    elif choice == 2:
        id = int(input("Enter Student ID to update: "))
        name = input("Enter New Name: ")
        age = int(input("Enter New Age: "))

        sql = "UPDATE student1 SET student_name=%s, age=%s WHERE id=%s"
        values = (name, age, id)

        mycursor.execute(sql, values)
        mydb.commit()

        print("Record updated successfully")

    # -------------------------------
    # DELETE
    # -------------------------------
    elif choice == 3:
        id = int(input("Enter Student ID to delete: "))

        sql = "DELETE FROM student1 WHERE id=%s"
        values = (id,)

        mycursor.execute(sql, values)
        mydb.commit()

        print("Record deleted successfully")

    # -------------------------------
    # DISPLAY
    # -------------------------------
    elif choice == 4:
        mycursor.execute("SELECT * FROM student1")
        result = mycursor.fetchall()

        print("\n--- Student Records ---")
        for row in result:
            print(row)

    # -------------------------------
    # EXIT
    # -------------------------------
    elif choice == 5:
        print("Exiting program...")
        break

    else:
        print("Invalid choice! Please try again.")

# Close connection
mycursor.close()
mydb.close()
```

## DDL Commands
---

```python
import mysql.connector

# Database connection
mydb = mysql.connector.connect(
    host="localhost",
    user="root",
    password="shiv_n0t_f0und",
    database="studentdb"
)

print("Database connected successfully")

mycursor = mydb.cursor()

while True:
    print("\n----- DDL MENU -----")
    print("1. Create Table")
    print("2. Alter Table")
    print("3. Drop Table")
    print("4. Show Tables")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    # -------------------------------
    # CREATE TABLE
    # -------------------------------
    if choice == 1:
        table_name = input("Enter table name: ")

        create_query = f"""
        CREATE TABLE {table_name} (
            id INT PRIMARY KEY,
            name VARCHAR(50),
            age INT,
            course VARCHAR(50)
        )
        """

        mycursor.execute(create_query)
        print("Table created successfully")

    # -------------------------------
    # ALTER TABLE
    # -------------------------------
    elif choice == 2:
        table_name = input("Enter table name to alter: ")

        print("1. Add Column")
        print("2. Modify Column")
        print("3. Drop Column")

        option = int(input("Enter choice: "))

        if option == 1:
            col_name = input("Enter column name: ")
            dtype = input("Enter datatype: ")
            query = f"ALTER TABLE {table_name} ADD {col_name} {dtype}"

        elif option == 2:
            col_name = input("Enter column name: ")
            query = f"ALTER TABLE {table_name} MODIFY {col_name} VARCHAR(100)"

        elif option == 3:
            col_name = input("Enter column name: ")
            query = f"ALTER TABLE {table_name} DROP COLUMN {col_name}"

        mycursor.execute(query)
        print("Table altered successfully")

    # -------------------------------
    # DROP TABLE
    # -------------------------------
    elif choice == 3:
        table_name = input("Enter table name to drop: ")

        query = f"DROP TABLE {table_name}"
        mycursor.execute(query)

        print("Table dropped successfully")

    # -------------------------------
    # SHOW TABLES
    # -------------------------------
    elif choice == 4:
        mycursor.execute("SHOW TABLES")

        print("\n--- Tables in Database ---")
        for table in mycursor:
            print(table)

    # -------------------------------
    # EXIT
    # -------------------------------
    elif choice == 5:
        print("Exiting program...")
        break

    else:
        print("Invalid choice!")

# Close connection
mycursor.close()
mydb.close()
```
