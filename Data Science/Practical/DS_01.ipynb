Difference Between Pandas and NumPy
1. NumPy:
       1. Used to work on Numerical data like 2d arrays and matrix. Here all the values are homogeneous i.e. all have same datatypes as numericals.
       2. Faster for numerical datasets under (50k rows) values.
2. Pandas:
       1. Used to work on Homogeneous Data i.e. Data having multiple datatypes as int, strings, etc.
       2. Better for very large tabular datasets (500k+ rows) requiring complex joins and grouping.
