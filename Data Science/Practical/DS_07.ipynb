```python
df["Date"]=pd.to_datetime(df["Date"])
df.head()
```

```python
plt.figure(figsize=(12,8))
plt.title("Simple Moving Average (Windows = 50)")
plt.xlabel("Date")
plt.ylabel("Close Price")
df["_50"]=df["Close Price"].rolling(window=50).mean()
plt.plot(df["Date"],df["_50"],color="green")
```

```python
plt.figure(figsize=(12,8))
plt.title("Simple Moving Average (Windows = 200)")
plt.xlabel("Date")
plt.ylabel("Close Price")
df["_200"]=df["Close Price"].rolling(window=200).mean()
plt.plot(df["Date"],df["_200"],color="blue")
```

```python
plt.figure(figsize=(12,8))
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.plot(df["Date"],df["Close Price"],color="red")
```

```python
plt.figure(figsize=(12,8))
plt.title("Simple Moving Average")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.plot(df["Date"],df["_50"],color="green")
plt.plot(df["Date"],df["_200"],color="blue")
plt.plot(df["Date"],df["Close Price"],color="red")
```
