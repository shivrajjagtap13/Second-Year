```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
img=cv.imread("lena.jpeg")
```

```python
cv.imwrite("img.jpg",img)
img1= cv.imread("img.jpg")
plt.imshow(img1)
```

```python
img_f=img.astype(np.float64)
r,g,b=cv.split(img_f)
cv.imwrite("img_f.jpg",img_f)
img_avg=(r+g+b)/3
cv.imwrite("avg_img.jpg",img_avg)
img2= cv.imread("avg_img.jpg")
plt.imshow(img2)
```

```python
luminosity=(0.29912*r)+(0.587*g)+(0.114*b)
cv.imwrite("lumi_img.jpg",luminosity)
img3= cv.imread("lumi_img.jpg")
lumi_img=plt.imshow(img3)
```

```python
luma=(0.2126*r)+(0.715*g)+(0.0722*b)
cv.imwrite("luma_img.jpg",luma)
img4= cv.imread("luma_img.jpg")
luma_img=plt.imshow(img4)
```

```python
lightness=(cv.max(r,g,b)+cv.min(r,g,b))/2
cv.imwrite("light_img.jpg",lightness)
img5= cv.imread("light_img.jpg")
light_img=plt.imshow(img5)
```

```python
import numpy as np
from math import log10, sqrt

def PSNR1 (I, I1):
    
    mse = np.mean((I - I1)**2)
    
    if mse == 0:
        return float('inf') #only possible if both the images are identical 
    max_pixel = 255.0
    psnr = 20 * log10(max_pixel / sqrt(mse)) 
    return psnr

print("PSNR of Original and Average:\t",PSNR1(img1,img2))
print("PSNR of Original and Lumi:\t",PSNR1(img1,img3))
print("PSNR of Original and Luma:\t",PSNR1(img1,img4))
print("PSNR of Original and Light:\t",PSNR1(img1,img5))
```

```python
from skimage.metrics import structural_similarity as ssim
```

```python
dr=255.0

score1=ssim(img_f,img2,channel_axis=2,data_range=dr)
score2=ssim(img_f,img3,channel_axis=2,data_range=dr)
score3=ssim(img_f,img4,channel_axis=2,data_range=dr)
score4=ssim(img_f,img5,channel_axis=2,data_range=dr)
print("SSIM Value of avg_img is:\t",score1)
print("SSIM Value of lumi_img is:\t",score2)
print("SSIM Value of luma_img is:\t",score3)
print("SSIM Value of light_img is:\t",score4)

#*score4=ssim(img_f,img5,channel_axis=2) what does channel axis means here and what is its use?
#In the skimage.metrics.structural_similarity (SSIM) function, the channel_axis parameter tells the algorithm which dimension of your image array represents the color channels (e.g., Red, Green, Blue). 
#Meaning of channel_axis=2
#In your specific code, channel_axis=2 indicates that the third dimension of your image array (index 2) contains the color information. 
#For example, if your image img_f has a shape of (Height, Width, 3), the index 2 points to the 3 (the color channels).
#Use of channel_axis
#Enables Multichannel Comparison: Standard SSIM is designed for 2D grayscale images. By setting channel_axis, you allow the function to handle color or multichannel data by calculating the SSIM for each channel independently and then averaging the results to give you a single score.
#Identifies Data Structure: Images can be stored in different formats.
#(H, W, C): Channels are at the end (common in OpenCV/Matplotlib). Here, channel_axis=2 or -1 is used.
#(C, H, W): Channels are at the beginning (common in PyTorch). Here, channel_axis=0 would be used.
#Prevents Errors: Without this parameter, if you pass a 3D color image, the function might mistake the color dimension for a third spatial dimension (like a 3D volume or depth), leading to incorrect calculations or a ValueError. 
#
#
#Automatic Guessing Issues: If data_range is not specified, skimage guesses it based on the image data type. For floating-point data, it defaults to a range of 2.0 (assuming a range from -1.0 to +1.0).
#Inaccurate Results: Most image processing uses a range of 0.0 to 1.0. Using the default data_range=2.0 on these images will result in incorrectly high SSIM values because the metric assumes the data spans a wider intensity range than it actually does.
#Recommendation: Documentation and developers recommend always passing data_range explicitly to ensure consistency, especially if your floating-point images represent values like 0–255 (where you would set data_range=255.0). 
```
